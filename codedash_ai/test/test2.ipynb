```python
import cv2
import time

def measure_video_fps(video_path):
    """
    Đo FPS của một video file
    """
    cap = cv2.VideoCapture(video_path)
    
    if not cap.isOpened():
        print(f"Không thể mở video: {video_path}")
        return None
    
    # Lấy FPS từ metadata của video
    fps_metadata = cap.get(cv2.CAP_PROP_FPS)
    
    # Đếm số frame thực tế
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    
    print(f"Video: {video_path}")
    print(f"FPS từ metadata: {fps_metadata:.2f}")
    print(f"Tổng số frame: {total_frames}")
    
    if fps_metadata > 0:
        duration = total_frames / fps_metadata
        print(f"Thời lượng video: {duration:.2f} giây")
    
    cap.release()
    return fps_metadata, total_frames

# Sử dụng
video_path = r"C:\Users\quang\Pictures\Her\Her\Video2.mov"

print("=== FPS METADATA ===")
measure_video_fps(video_path)
```
